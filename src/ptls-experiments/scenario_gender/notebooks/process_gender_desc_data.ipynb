```python
import pandas as pd
```

```python
pd_tr_mcc_codes = pd.read_csv('../data/mcc_translated.csv', sep=',')
pd_tr_types = pd.read_csv('../data/tr_type_translated.csv', sep=',')
pd_tr_mcc_codes, pd_tr_types
```

```python
import json

with open('../data/mcc_code_mapping.json', 'r') as f:
    mcc_code_mappings = json.load(f)
    orig_mcc_code_to_new = {v: k for k, v in mcc_code_mappings.items()}

with open('../data/tr_type_mapping.json', 'r') as f:
    tr_type_mappings = json.load(f)
    orig_tr_type_to_new = {v: k for k, v in tr_type_mappings.items()}
```

```python
pd_tr_types
```

```python
import numpy as np 

mcc_dict = dict(zip(pd_tr_mcc_codes['mcc_code'], pd_tr_mcc_codes['mcc_description']))

mcc_dict = dict()
for row in pd_tr_mcc_codes.itertuples():
    orig_mcc_code = str(row.mcc_code)
    new_mcc_code = orig_mcc_code_to_new[orig_mcc_code]
    mcc_dict[new_mcc_code] = row.mcc_description

type_dict = dict()
for row in pd_tr_types.itertuples():
    orig_tr_type = str(row.tr_type)
    
    if orig_tr_type not in orig_tr_type_to_new:
        continue

    new_tr_type = orig_tr_type_to_new[orig_tr_type]
    type_dict[new_tr_type] = row.tr_description

len(mcc_dict), len(type_dict)
```

```python
# save the mcc_dict as mcc_dict.npy
np.save('../data/mcc_dict.npy', mcc_dict)
# save the type_dict as type_dict.npy
np.save('../data/type_dict.npy', type_dict)
```
